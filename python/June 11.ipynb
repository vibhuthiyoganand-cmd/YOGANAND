### Using `strftime` to convent datetime values to string

- We can format the datetime values as string
- We use `strftime()` method, where `f` f stands for `format`
#### *syntax*

```
datetime_object.strftime(format_str)
```

#### `format_str` codes
```
%y - two digit year number
%Y - four digit year number
%m - month number
%d - day number of the month
%H - hour number (24-hours format)
%I - hour number(12-hours format)
%M - minute number
%S - secound number
%p - AM/PM Specifier
%a - short weekday name(sun,mon,tue,...)
%A - full weekday name(sunday,monday,tueday,...)
%b - short month name(jan,fed,mar,...)
%B - full month name(january,fedruary,march,...)



````python
one_day = datetime (2025, 3, 2, 12, 35, 30)
print(one_day)
````

````python
one_day.strftime("(%A) %b %d, %Y")
````

### The `timedelta` object-span of time

- The `datetime ` modules contian a `timedelta constructor to create time span
- These time spans are internally difference between two value

````python
# importing timedelta contructor
from datetime import timedelta 
````

###  **syntax**
```
timedelta([weeks,] [days,] [hours,] [mintes,] [second,] [millisecond,] [microsecond])

```

````python
project_duration = timedelta(weeks=6)
start_date = date(2026, 6, 15)
end_date = start_date + project_duration
print(end_date)
````

````python
task_span = timedelta(hours=3, minutes=30)
start_time = datetime (2026, 6, 11, 12, 5)
end_time = start_time + task_span
print(end_time.strftime("End time: %I:%M;%S %P"))
````
