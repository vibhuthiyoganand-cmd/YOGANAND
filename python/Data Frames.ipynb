# Introduction to Data Frames

- A data frames represent a tablear data (table)
- Contains rows and colums
- By colums are also referred by their index(0 to n-1)
- And colums are also referred by their index
- Rows are axis=0 (x-axis) and colums are at axis(y-axis)

````python
# imports
import numpy as np
import pandas as pd
````

````python
data = {
    'city':['hyd','del','mum','che','kol'],
    'state':['telanagana','delhi','maharastra','tamilnadu','west bengal'],
    'population':[1500000, 1800000, 1400000, 16000000, 1200000]
}
````

````python
pd.DataFrame(data)
````

````python
customer_visits = np.random.randint(10, 20,(3,5))
customer_visits
````

````python
pd.DataFrame(customer_visits)
````

````python
pd.DataFrame(customer_visits,
             index=['morning','afternoon','evening'],
             columns=['mon','tue','wed','thu','fri'])
````
