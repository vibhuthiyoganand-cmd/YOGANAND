## Local and global variables

### Global variable
- global variables are declared/created outside of any function
- They are accessblie outside of a function and also inside any function in the programm
- By defult, inside function the global variable are read-only
- We can modify a global variable inside a function using the `global` keyword however, this is not recommended


### Local variable
- A local variable is created/declared inside a function
- It is only accessible inside the particular function
- *Note:* If a local and global variables have same name, then inside it is always local that ia accessed,outside it is always global variable that is accessed


````python
def sample():
    print(x)   # global variable

x = 25   # global variable
sample()
````

````python
def sample():
    x = 10
    print(x)   # local variable

x = 50   # global variable
sample()
print(x)
````

````python
def sample(x):  # local
    print(x)    # local 

x = 25   # global variable
sample(x)
print(x)
````

````python
def sample(x):  # local
    x = x * 3
    print(x)    # local 

x = 30   # global variable
sample(x)
print(x)
````

````python
def sample():  
    global x     # global - write moda; not recommended
    x = x * 10   
    print(x)

x = 5  # global 
sample()
print(x)
````

````python
val = random.randrange(1, 10)
attempts = 0

while attempts < 3:
    guess = int(input("Enter the guess: "))

    if guess > val:
        print("Your guess is greater than the value.")
    elif guess < val:
        print("Your guess is less than the value.")
    else:
        print("Your guess is right!!")
        break

    attempts += 1

print("The value was:", val)
print("Wrong guesses:", attempts)
````
