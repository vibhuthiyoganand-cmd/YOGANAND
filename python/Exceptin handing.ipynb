### Errors vs excepetion

#### Errors(syntax)
- when?
  - before the program runs
- why?
  - The code is grammatically incorrect for python
- Can we handle?
  - No-fix the code
- Example
  - print("hello")
  - (missing parenthesess)

#### Exceptions
- When?
  - During execution (runtime)
- Why?
  - The code is vaild, but something unexpected happend
- Can we handle ?
  - Yes-catch and reposend
- Example:
  - 1/0 (ZeroDivisionError)

**Note**: When an excepetion occurs python an `exception object` and  "raises it .if we don't catch  the expcetion the program terminates with a traceback

### The `try/except` mechanism

- We use `try` and `except` to tell python: "try this block, and if that type of problem occurs , do this other thing instead."

#### Basic structure
```
try:
    # code that might go wrong
    risky_operation()
except someException:
    # What to do if that specifi problem occur 
    recovery_plan()
     

````python
# Vending machine simulation

```
inventory = {"chips":0, "cola":5}

def dispense(item):
    if inventory[item] ==0:
        raise OutOfStockError(item)
        inventory[item]-=1
        print(f"Enjoy your {item}!")

try:
    dispence("chips")
except OutOfStockError:
    print("sorry, that the item is not available.money returned:")
```
````

````python
# Catching ZeroDivisionError
try:
    numerator = 10
    denominator = 0
    result = numerator/denominator
    print("result:", result)
except ZeroDivisionError:
    print("Oops! you can't divide by zero:")

print("program continues...")    
````

#### Catching specific Exceptions
- multiple `except` block, exception hierarchy, ordering 

````python
# Handing different error types separately
try:
    value = int(input("Enter a number: "))
    result = 100 / value
except ValueError:
    print("That wasn't a vaild number!")
except ZeroDivisionError:
    print("Cannot divide by zero!")

print("Carring on..")
````
