# working with dictinaries

- A collection of items, unlike list items, they are unordered
- Each item in dictinaries  contian `key` and `value` pairs
- In one sence , the key word as index (in fact , no indexing here)
- The values are union in a dictinaries , but valuse can repect.
- The key values are of immutable types:int,float,str,tuple

### **syntax**
``` 
(key1:value1, key2:value2, key3:value3, key4:value,,...)

```

### Looping through keys

````python
for k in countries:
    print(k, end=' ')
````

````python
# optionally we can also use - keys() method to get key values
for k in countries.keys():
    print(k, end=' ')
````

````python
### Looping throgh using values
for k in countries.values():
    print(k, end=' ')
````


### Looping throgh items using values () method

- returnes a truple of key and values pairs for each item


````python
for k in countries.items():
    print(k, end=' ')
````

````python
for item in countries.items():
    print(item)
````
