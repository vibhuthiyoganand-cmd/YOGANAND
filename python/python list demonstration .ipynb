### THE PHOTO ALBUM - SLICING & DICING

```python
print("=" * 60)
```

```python
print("slicing subset for workshop")
```

```python
print("=" *60)
```

```python
# we'll use the cuurent attendees list(already sorted? no, we'll sort later) 
```

```python
workshop_A = attendees [:5]    # first 5
```

```python
workshop_B = attendees [3:8]  # indices 3,4,5,6,7
```

```python
workshop_no_ends = attendees [1:-1] # remove first and last
```

```python
reverse_list = attendees [::-1] # reversed for a display board 
```

```python
every_other = attendees [::2]   # every secound attendees 
```

```python
print("workshop A (first 5):", [a[0] for a in workshop_A])
```

```python
print("workshop B (3::8):", [a[0] for a in workshop_B])
```

```python
print("without first/last:", [a[0] for a in workshop_no_ends])
```

```python
print("reversed:", [a[0] for a in reverse_list])
```

```python
print("every other:", [a[0] for a in every_other])
```
