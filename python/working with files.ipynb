## text files

#### 3 files operations
1. open file - open() function
2. Read from the files or write to the file. read()/write()method
3. close file-close()method


### the built in open(file name,mode)function

- `file name` : its a filename or filepath +filename
- `mode` : there are different modes in which we can open a files
- the open () function create an file object when a files opened: we can use that file object to work with the file and perform another file operations

#### **modes**

- 'r' (read )   using this mode we can read an existing files. If the file doesn't exists it throws 'Files no foundError
- 'w' (write) - using this mode if the files doesn't exists then it creates If the files alerady exists then it erases all the data from it.
- 'a' (append) - using this mode , if the files doesn't exists then it create one. if the files already exists then it append the new to the end of the files
- 'b' (binary) - to work with binary files:'rb' for reading and 'wb' writing
