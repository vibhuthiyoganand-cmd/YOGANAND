### 23 . ` format(dataitem(s))` method

#### **syntax**
```
"...{:[width][comma][.deci_place][types]}...".format(datatime)

#### `[width][comma][.deci_place][types]`-format specification
- `width` - optional; specifiy the width foe datatime
- `comma` - optional; a number can contain comma(thousand separtor)
- `deci_place` - optional; how many decimal place are needed
- `types` - tyoes of data you are formating :d(for integer),f(for float),e(for scientific notation)...
- `<`for right-alignment,`.>` for left-aligment^ for center

`````python
"{:10}".format("hello")
`````

`````python
"{:15}".format(123)
`````

`````python
"{:20,}".format(123456789)
`````

`````python
"{:20,.2f}".format(12345.67890987)
`````

`````python
"{:<20}".format(1000)
`````

`````python
"{:>20}".format(1000)
`````

`````python
"{:>20}".format("hello")
`````

`````python
"{:^30}".format(12345)
`````

"{:20.4e}".format(123456789)

`````python
print("{:25}{:>6}{:>20}".format("product name","units","price"))
print("{:25}{:>6}{:>20}".format("------------","-----","-----"))
print("{:25}{:6}{:20,.2f}".format("Hp Ryzen 5",10,48000.0))
print("{:25}{:6}{:20,.2f}".format("Epson projext",2,32000.0))
print("{:25}{:6}{:20,.2f}".format("printer",7,18250.0))
print("{:25}{:6}{:20,.2f}".format("iphone 17",8,180000.0))
print("{:25}{:6}{:20}".format("samsung A35",7,32000.0))
`````

`````python
"My salary is {:,.3f} rupees.".format(100000.0)
`````
