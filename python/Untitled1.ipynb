# SET METHOD

### 1.`additem()` method

- This method add an specified item a set

```python
print(colours)
```

```python
colours.add("orange")
print(colours)
```

### 2.`updateitem()` method

- We can an more than one item using update method

```python
colours.update(["pink", "voilet", "cyan",])
print(colours)
```

### `removeitem()` method

- To remove the specified item set 

```python
colours.remove("red")
print(colours)
```

```python
colours.remove("green")
print(colours)
```

### 1. `unionset()` method

```python
a = {1,2,3,4,}
b = {3,4,5,6,}
```

```python
a.union(b)
a|b
```

 ### `intersection(set)` or '&' intersection operations]

```python
a.intersection(b)
a&b
```

### 3.`difference between sets`

```python
 a-b
 b-a
```

### 4. `symmetric_difference(set)` operator
- find the symmetric diiference between to sets

```python
a.symmetric_difference(b)
a^b
```
