## the `for` loop construct
- collection (count of collection) based loop
- the loop execute the block for each item in the collection.
- Each executiobn is called an iteration.
- And a collection never has infinite number of items.

#### syntax

```
for <var> in colleaction:
block

```

- the loop iteration each item in collection 
- In each iteration it assigns the iteam to the var, and then execute the block

## the `break` statement
- it work for both the `for` and `while` loop equally
- the`break` statement ends the loop early; without going through all the iteration
- alwats ues the `break` statement inside the `if`statement , because you break a loop for a reson

`````python
attendess = ['shubham' , 'sherya' , 'nandu' , 'risitha' , 'prakash' , 'hasini']
for e in attendess:
    print(e)
    if e=='prakesh':
        break 
`````

`````python
value = [ 100, 88, 43, 59, 49, 59, 99, 300, -15]
for i in value:
    if i % 7== 0:
        print(i,"Divisible by 7.")
        break
`````

### the `continue` staement
- To skip an iteration we use `continue` statement
- We write the `continue` statement the `if` to skip an iteration for a reson
- When the `continue` statement is excuted the control jump to the beginning to the loop to for next iteratin

`````python
weekdays = ['monday' , 'tuesday' , 'wednesday' , 'thursday' , 'friday' , 'saturday' , 'sunday']
# skip printing days starting with 's' 0s 't' , print remaing
for day in weekdays:
    if day[0] == 's' or day[0] == 't':
        continue 
        print(day)
`````
